```python
# review null to avg median(because median doesnt get affected by outliers)
#review rating within each category and then replace with null
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
```

```python
df.columns
```

```python
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
#create column age group
labels=['Young adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# Corrected mapping dictionary with proper spelling and formatting
frequency_mapping = {
    'Weekly': 7,
    'Fortnightly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Annually': 365
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
# 1. Configure the local PostgreSQL server (set password and create the database)
!sudo -u postgres psql -c "ALTER USER postgres PASSWORD 'Pranjal@46';"
!sudo -u postgres psql -c "CREATE DATABASE customer_behavior;"

# 2. Connect and load the data using SQLAlchemy
from sqlalchemy import create_engine
import urllib.parse

username = 'postgres'
password = 'Pranjal@46'
host = 'localhost'
port = '5432'
database = 'customer_behavior'

# Safely URL-encode password containing special characters
safe_password = urllib.parse.quote_plus(password)

engine = create_engine(f'postgresql+psycopg2://{username}:{safe_password}@{host}:{port}/{database}')

# Load DataFrame into the table
table_name = 'customer'
df.to_sql(table_name, engine, if_exists='replace', index=False)
print(f"Data successfully loaded into table '{table_name}' in database '{database}'.")
```

```python
from sqlalchemy import inspect

# Create an inspector inspector object
inspector = inspect(engine)

# Get schema names
schemas = inspector.get_schema_names()
print(f"Available schemas: {schemas}")

# Get table names in the default 'public' schema
tables = inspector.get_table_names()
print(f"Tables in default schema (public): {tables}")
```

```python
import pandas as pd
from sqlalchemy import text

# Query the database using text() wrapper to verify the row count
with engine.connect() as connection:
    row_count = connection.execute(text("SELECT COUNT(*) FROM customer;")).scalar()
    print(f"Total rows in the 'customer' table: {row_count}")

    # Display first few rows directly from the SQL database
    db_df = pd.read_sql(text("SELECT * FROM customer LIMIT 5;"), connection)
    display(db_df)
```
